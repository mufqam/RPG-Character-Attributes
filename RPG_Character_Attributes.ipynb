```python
import pandas as pd
data1 = pd.read_csv('/content/archive.zip')
data1
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 2D Histogram of Class vs FBoss
df_2dhist = data1.pivot_table(index='Class', columns='FBoss', aggfunc=len, fill_value=0)
fig, ax = plt.subplots(figsize=(12, 7))
im = ax.imshow(df_2dhist, cmap='Greens')

# Show all ticks and label them with the respective list entries
ax.set_xticks(np.arange(len(df_2dhist.columns)), labels=df_2dhist.columns)
ax.set_yticks(np.arange(len(df_2dhist.index)), labels=df_2dhist.index)

# Create colorbar
cbar = ax.figure.colorbar(im, ax=ax, cmap='Greens')
cbar.ax.set_ylabel('Counts', rotation=-90, va="bottom")

# Loop over data dimensions and create text annotations.
for i in range(len(df_2dhist.index)):
    for j in range(len(df_2dhist.columns)):
        text = ax.text(j, i, df_2dhist.iloc[i, j],
                       ha="center", va="center", color="w")

ax.set_title("2D Histogram of Class vs FBoss")
fig.tight_layout()
plt.show()


# Histograms for each numerical column
for column in data1.select_dtypes(include=np.number).columns:
    plt.figure(figsize=(10, 5))
    plt.hist(data1[column], bins=20, color='skyblue', edgecolor='black')
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.grid(axis='y', alpha=0.75)
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Scatter plots for pairs of numerical columns
numerical_cols = data1.select_dtypes(include=np.number).columns
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=data1, x=numerical_cols[i], y=numerical_cols[j], hue='FBoss', alpha=0.7)
        plt.title(f'Scatter Plot of {numerical_cols[i]} vs {numerical_cols[j]}')
        plt.grid(True)
        plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Box plots for each numerical column by Class
numerical_cols = data1.select_dtypes(include=np.number).columns
for col in numerical_cols:
    plt.figure(figsize=(12, 7))
    sns.boxplot(data=data1, x='Class', y=col, hue='FBoss')
    plt.title(f'Box Plot of {col} by Class')
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.show()
```


**Distribusi:**

*   **Class vs FBoss:** Dari histogram 2D, kita dapat melihat jumlah karakter dari setiap kelas yang telah dan belum mengalahkan bos terakhir (FBoss). Ini memberikan gambaran tentang kelas mana yang lebih berhasil dalam mencapai tujuan akhir permainan.
*   **Distribusi Atribut Numerik:** Histogram untuk setiap atribut numerik (Armor, Weapon, Physical, Magic, Level) menunjukkan bagaimana nilai-nilai ini didistribusikan di seluruh kumpulan data. Kita dapat melihat apakah sebagian besar karakter memiliki statistik rendah atau tinggi di area tertentu.

**Hubungan:**

*   **Scatter Plots:** Plot pencar menunjukkan hubungan antara pasangan atribut numerik. Misalnya, kita dapat melihat apakah ada korelasi antara level karakter dan kekuatan senjata mereka. Pewarnaan berdasarkan 'FBoss' membantu kita melihat apakah ada kombinasi atribut tertentu yang umum di antara karakter yang telah mengalahkan bos terakhir.
*   **Box Plots:** Plot kotak memberikan ringkasan statistik (median, kuartil, pencilan) untuk setiap atribut numerik, yang dikelompokkan berdasarkan kelas karakter. Ini memungkinkan perbandingan langsung antar kelas. Misalnya, kita bisa melihat apakah kelas 'Magician' secara konsisten memiliki statistik 'Magic' yang lebih tinggi daripada kelas 'Warrior'.

Secara keseluruhan, visualisasi ini memberikan wawasan tentang karakteristik berbagai kelas karakter dalam game dan faktor-faktor yang mungkin berkontribusi pada keberhasilan mereka dalam mengalahkan bos terakhir.

```python
# Calculate and display the correlation matrix
correlation_matrix = data1.select_dtypes(include=np.number).corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Prepare the data
X = data1.select_dtypes(include=np.number)
y = data1['FBoss']

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Create and train the logistic regression model
model = LogisticRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
print('\\nClassification Report:')
print(classification_report(y_test, y_pred))
```

```python
from scipy.stats import ttest_ind

# Separate the 'Level' data for players who have and haven't defeated the final boss
fboss_true_level = data1[data1['FBoss'] == True]['Level']
fboss_false_level = data1[data1['FBoss'] == False]['Level']

# Perform the t-test
ttest_result = ttest_ind(fboss_true_level, fboss_false_level)

# Print the results
print(f"T-test results for 'Level' between FBoss True and False groups:")
print(f"T-statistic: {ttest_result.statistic:.2f}")
print(f"P-value: {ttest_result.pvalue:.3f}")

# Interpret the results
if ttest_result.pvalue < 0.05:
    print("The difference in mean 'Level' is statistically significant.")
else:
    print("The difference in mean 'Level' is not statistically significant.")
```

*   **Matriks Korelasi:** Analisis ini menunjukkan hubungan linear antara berbagai atribut numerik. Misalnya, kita dapat melihat korelasi positif yang kuat antara 'Level' dan 'Magic', yang menunjukkan bahwa karakter dengan level lebih tinggi cenderung memiliki statistik sihir yang lebih tinggi.

*   **Regresi Logistik:** Model regresi logistik yang kami buat memiliki akurasi yang sangat tinggi (100%) dalam memprediksi apakah seorang pemain telah mengalahkan bos terakhir ('FBoss') berdasarkan atribut numerik mereka. Ini menunjukkan bahwa atribut-atribut seperti 'Level', 'Magic', 'Physical', 'Weapon', dan 'Armor' sangat prediktif terhadap keberhasilan dalam permainan.

*   **Uji-t:** Hasil uji-t menunjukkan bahwa ada perbedaan yang signifikan secara statistik dalam rata-rata 'Level' antara pemain yang telah mengalahkan bos terakhir dan yang belum. P-value yang sangat rendah (0.000) mengkonfirmasi bahwa perbedaan ini bukan karena kebetulan. Ini memperkuat gagasan bahwa level karakter adalah faktor penting untuk keberhasilan.

Secara keseluruhan, analisis statistik ini menunjukkan bahwa atribut karakter, terutama level, memiliki hubungan yang kuat dengan kemampuan untuk menyelesaikan permainan.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create and train the Random Forest Classifier model
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_rf = rf_model.predict(X_test)

# Evaluate the model
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print(f'Random Forest Accuracy: {accuracy_rf:.2f}')
print('\\nRandom Forest Classification Report:')
print(classification_report(y_test, y_pred_rf))
```

Model *Random Forest Classifier* yang kita buat berhasil memprediksi status kekalahan bos terakhir (*FBoss*) dengan akurasi sekitar 93%. Ini berarti model tersebut benar dalam sekitar 93 dari 100 kasus pada data uji.

Laporan klasifikasi memberikan rincian lebih lanjut:
- **Precision (Presisi):**
  - Untuk kelas `False` (belum mengalahkan bos), presisinya adalah 95%. Artinya, dari semua prediksi `False` yang dibuat model, 95% di antaranya benar.
  - Untuk kelas `True` (sudah mengalahkan bos), presisinya adalah 89%.
- **Recall (Daya Ingat):**
  - Untuk kelas `False`, recall-nya adalah 92%. Artinya, dari semua kasus `False` yang sebenarnya, model berhasil mengidentifikasi 92% di antaranya.
  - Untuk kelas `True`, recall-nya adalah 93%.
- **F1-Score:**
  - F1-score adalah rata-rata harmonik dari presisi dan recall. Skor ini memberikan gambaran yang seimbang tentang kinerja model, terutama jika ada ketidakseimbangan antar kelas.

Secara keseluruhan, model *Random Forest* ini menunjukkan kinerja yang sangat baik dalam memprediksi keberhasilan pemain, meskipun sedikit lebih rendah dari model Regresi Logistik yang kita buat sebelumnya (yang mencapai akurasi 100%). Perbedaan ini bisa jadi karena *Random Forest* lebih tahan terhadap *overfitting* dan mungkin memberikan gambaran yang lebih realistis tentang bagaimana model akan bekerja pada data baru yang belum pernah dilihat sebelumnya.
